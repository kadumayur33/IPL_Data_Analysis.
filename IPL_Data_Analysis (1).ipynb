```python
import pandas as pd

deliveries = pd.read_csv("https://raw.githubusercontent.com/dsrscientist/dataset1/master/IPL_Ball_by_Ball_2008_2020.csv")
matches = pd.read_csv("https://raw.githubusercontent.com/dsrscientist/dataset1/master/IPL_Matches_2008_2020.csv")

deliveries.head()
matches.head()
```

# **Task 1 — Grouping, Filtering with .isin(), and Strike Rate Calculation**

```python
# Filter overs 16–20
death_overs = deliveries[(deliveries['over'] >= 16) & (deliveries['over'] <= 20)]

# Group by batsman
grouped = death_overs.groupby('batsman').agg(
    total_runs=('batsman_runs', 'sum'),
    balls_faced=('ball', 'count')
).reset_index()

# Strike rate
grouped['strike_rate'] = (grouped['total_runs'] / grouped['balls_faced']) * 100

# Filter batsmen with >=200 balls
qualified_batsmen = grouped[grouped['balls_faced'] >= 200]['batsman'].tolist()

filtered = grouped[grouped['batsman'].isin(qualified_batsmen)]

# Sort and top 10
top_10 = filtered.sort_values(by='strike_rate', ascending=False).head(10)

print(top_10)
```

## **Task 2 — Merging DataFrames and Multi-Column Grouping**

```python
# Merge datasets
merged = pd.merge(deliveries, matches, on='id', how='left')

# Group by season and batsman
season_runs = merged.groupby(['season', 'batsman'])['batsman_runs'].sum().reset_index()

# Sort
season_runs = season_runs.sort_values(['season', 'batsman_runs'], ascending=[True, False])

# Get top scorer per season
orange_cap = season_runs.drop_duplicates(subset=['season'], keep='first')

print(orange_cap)
```

# **Task 3 — Handling Missing Values**

```python
# Missing values count
print(matches.isnull().sum())

# Drop rows where all values are null
matches = matches.dropna(how='all')

# Fill missing values
matches['city'] = matches['city'].fillna('Unknown')
matches['winner'] = matches['winner'].fillna('No Result')

# Verify
print(matches['city'].isnull().sum())
print(matches['winner'].isnull().sum())

# Value counts
print(matches['city'].value_counts())
```

```python
#explaination:
We use fillna() instead of dropna() for columns like city and winner because these values may be missing but the match data is still important. Dropping rows would result in loss of useful data. Using placeholder values like "Unknown" and "No Result" helps preserve the dataset while still handling missing information properly.
```
